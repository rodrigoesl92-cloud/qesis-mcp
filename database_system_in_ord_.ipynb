El archivo de la ontología no es un archivo CSV típico. Parece ser un formato de texto personalizado que define clases y sus propiedades. Para procesar esto, podemos leer el archivo como un texto completo y luego usar expresiones regulares para extraer la información estructurada.

````
import re

# Leer el contenido completo del archivo
with open('ASO_Digital_Twin_Ontology_Edition(Ontology_Code) (1).csv', 'r') as f:
    ontology_text = f.read()

# Imprimir el contenido para verificar
print("Contenido del archivo de ontología:")
print(ontology_text)
````

Ahora podemos usar expresiones regulares para identificar las clases y sus propiedades. Por ejemplo, podemos buscar líneas que comiencen con `class ` para encontrar los nombres de las clases y luego las líneas indentadas para encontrar las propiedades.

````
ontology_data = {}
current_class = None

# Strip leading/trailing whitespace, including newlines
ontology_text_to_parse = ontology_text.strip()

# Eliminar comillas dobles si el contenido completo está envuelto en ellas
if ontology_text_to_parse.startswith('"') and ontology_text_to_parse.endswith('"'):
    ontology_text_to_parse = ontology_text_to_parse[1:-1]

# Dividir el texto en líneas y procesar cada una
for line in ontology_text_to_parse.split('\n'):
    line = line.strip()
    if not line:
        continue

    # Buscar el inicio de una nueva clase
    class_match = re.match(r'class\s+([A-Za-z_][A-Za-z0-9_]*):', line)
    if class_match:
        current_class = class_match.group(1)
        ontology_data[current_class] = {'properties': []}
    elif current_class:
        # Asumir que las líneas siguientes (indentadas o no) son propiedades de la clase actual
        # Aquí necesitamos una lógica más sofisticada si la indentación es clave o si hay subClases, etc.
        # Por ahora, simplemente tomamos la línea como una propiedad.
        # Consideramos propiedades si no son 'subClassOf' o similares, que podrían ser relaciones.
        if '=' in line or '[' in line: # Simple heuristic to identify properties vs. relations
             ontology_data[current_class]['properties'].append(line)
        else:
            # If it's a simple property name, just add it
            ontology_data[current_class]['properties'].append(line.split(' ')[0])

# Imprimir la estructura de datos extraída
print("Estructura de la ontología extraída:")
for cls, data in ontology_data.items():
    print(f"  Clase: {cls}")
    for prop in data['properties']:
        print(f"    - Propiedad: {prop}")
````

````
import pandas as pd

# Preparar los datos para el DataFrame
data_for_df = []
for class_name, class_info in ontology_data.items():
    for prop in class_info['properties']:
        data_for_df.append({'Class': class_name, 'Property': prop})

# Crear el DataFrame
df_ontology_structured = pd.DataFrame(data_for_df)

# Mostrar el DataFrame
display(df_ontology_structured)
````

## Explicar Coincidencia Exacta vs. Parcial

### Subtask:
Detallar la diferencia entre la coincidencia exacta y la coincidencia parcial (o difusa), incluyendo ejemplos y sus implicaciones para la integración de datos.


### Coincidencia Exacta (Exact Matching)

La **coincidencia exacta** se refiere a la integración de datos donde los valores de una columna en un DataFrame deben ser **idénticos** a los valores de una columna en otro DataFrame para que se considere una "coincidencia". Esto significa que cada carácter, incluyendo mayúsculas, minúsculas, espacios y puntuación, debe ser exactamente igual.

**Ejemplo:**

Si tenemos `df_executive_summary['Metric_Pillar'] = 'AI Governance'` y `df_ontology_structured['Property'] = 'AI Governance'`, esto se consideraría una coincidencia exacta. Sin embargo, si `df_ontology_structured['Property']` fuera `'AI Governance '` (con un espacio extra) o `'AI-Governance'`, **no** sería una coincidencia exacta.

**Implicaciones:**
*   **Alta Precisión**: Cuando se encuentra una coincidencia, es muy probable que sea correcta, ya que los valores son idénticos.
*   **Baja Recuperación (Recall)**: Es muy estricta y puede pasar por alto muchas relaciones legítimas debido a pequeñas variaciones (errores tipográficos, abreviaturas, sinónimos, diferencias en el formato).
*   **Eficiencia**: Generalmente es muy rápida y computacionalmente económica de realizar, especialmente con índices de bases de datos.

### Coincidencia Parcial o Difusa (Partial / Fuzzy Matching)

### Coincidencia Exacta (Exact Matching)

La **coincidencia exacta** se refiere a la integración de datos donde los valores de una columna en un DataFrame deben ser **idénticos** a los valores de una columna en otro DataFrame para que se considere una "coincidencia". Esto significa que cada carácter, incluyendo mayúsculas, minúsculas, espacios y puntuación, debe ser exactamente igual.

**Ejemplo:**

Si tenemos `df_executive_summary['Metric_Pillar'] = 'AI Governance'` y `df_ontology_structured['Property'] = 'AI Governance'`, esto se consideraría una coincidencia exacta. Sin embargo, si `df_ontology_structured['Property']` fuera `'AI Governance '` (con un espacio extra) o `'AI-Governance'`, **no** sería una coincidencia exacta.

**Implicaciones:**
*   **Alta Precisión**: Cuando se encuentra una coincidencia, es muy probable que sea correcta, ya que los valores son idénticos.
*   **Baja Recuperación (Recall)**: Es muy estricta y puede pasar por alto muchas relaciones legítimas debido a pequeñas variaciones (errores tipográficos, abreviaciones, sinónimos, diferencias en el formato).
*   **Eficiencia**: Generalmente es muy rápida y computacionalmente económica de realizar, especialmente con índices de bases de datos.
